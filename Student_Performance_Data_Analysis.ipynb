```python
import pandas as pd

df = pd.read_csv('/content/StudentPerformanceFactors.csv')

print("Dataset Loaded Successfully!")
print(df.head())
```

```python
print("Number of Rows and Columns:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

```python
df = df.drop_duplicates()

for col in df.columns:
    if df[col].isnull().sum() > 0:
        if df[col].dtype == 'object':
            df[col] = df[col].fillna(df[col].mode()[0])
        else:
            df[col] = df[col].fillna(df[col].median())

print("Data Cleaning Completed!")

print("Remaining Missing Values:", df.isnull().sum().sum())
print("Remaining Duplicate Rows:", df.duplicated().sum())
```

```python
print("Statistical Analysis:")
print(df.describe())
```

```python
print(df.columns.tolist())
```

```python
print("Student Performance Analysis")

print("\nAverage Study Hours:")
print(df['Hours_Studied'].mean())

print("\nAverage Attendance:")
print(df['Attendance'].mean())

print("\nAverage Sleep Hours:")
print(df['Sleep_Hours'].mean())

print("\nAverage Previous Scores:")
print(df['Previous_Scores'].mean())

print("\nAverage Exam Score:")
print(df['Exam_Score'].mean())

print("\nHighest Exam Score:")
print(df['Exam_Score'].max())

print("\nLowest Exam Score:")
print(df['Exam_Score'].min())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))
plt.hist(df['Exam_Score'], bins=10)
plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")
plt.show()


plt.figure(figsize=(7, 5))
plt.scatter(df['Hours_Studied'], df['Exam_Score'])
plt.title("Study Hours vs Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.show()


average_score = df.groupby('Parental_Involvement')['Exam_Score'].mean()

plt.figure(figsize=(7, 5))
average_score.plot(kind='bar')
plt.title("Parental Involvement vs Exam Score")
plt.xlabel("Parental Involvement")
plt.ylabel("Average Exam Score")
plt.show()
```
